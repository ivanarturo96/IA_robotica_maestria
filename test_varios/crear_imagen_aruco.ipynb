```python
import cv2
import numpy as np

# Cargar el diccionario ArUco
diccionario = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_50)

# Elegir el ID del marcador (0 a 49 en este diccionario) y el tamaño en píxeles
id_marcador = 23
tamano_px = 200

# Generar la imagen del marcador
imagen_marcador = np.zeros((tamano_px, tamano_px), dtype=np.uint8)
cv2.aruco.generateImageMarker(diccionario, id_marcador, tamano_px, imagen_marcador, 1)

# Guardarla como archivo
cv2.imwrite("marcador_aruco_23.png", imagen_marcador)
```
